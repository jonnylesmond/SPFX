```python
import seaborn
import pandas
import matplotlib.pyplot as plt
```

```python
trainingdata, walesdata = (
    pandas.read_csv('aggdata/aggpops_transposed(Sheet1).csv'),
    pandas.read_csv('WalesData/WalesPops-Transposed.csv')
)

trainingdata
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12,8))

seaborn.lineplot(data=walesdata, x='Date', y='Cardiff', ax=axes[0,0])
seaborn.lineplot(data=walesdata, x='Date', y='Swansea', ax=axes[0,1])
seaborn.lineplot(data=walesdata, x='Date', y='Newport', ax=axes[1,0])
seaborn.lineplot(data=walesdata, x='Date', y='Blaenau Gwent', ax=axes[1,1])
```
